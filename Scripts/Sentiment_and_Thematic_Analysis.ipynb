```python
# NEW CELL 1: Import and Run Sentiment/Themes
import pandas as pd
df = pd.read_csv("../data/processed/reviews_processed.csv")


from sentiment_themes import SentimentThemesAnalyzer
analyzer = SentimentThemesAnalyzer()
success = analyzer.analyze()
if success:
    df_sent = analyzer.df  # Enhanced DF
    print("Task 2 complete! Coverage:", analyzer.stats['sentiment_coverage'])
```

```python
# NEW CELL 2: Visualize Sentiment by Bank
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(data=df_sent, x='bank_name', y='sentiment_score', hue='rating')
plt.title('Sentiment Scores by Bank and Rating')
plt.xticks(rotation=45)
plt.show()
```

```python
# NEW CELL 3: Theme Distribution
plt.figure(figsize=(12, 6))
theme_explode = df_sent['identified_themes'].value_counts().head(10)
plt.pie(theme_explode.values, labels=theme_explode.index, autopct='%1.1f%%')
plt.title('Top Themes Across All Reviews')
plt.show()
```

```python
# NEW CELL 4: Per-Bank Themes Table
print("Themes per Bank:")
for bank in df_sent['bank_name'].unique():
    bank_df = df_sent[df_sent['bank_name'] == bank]
    print(f"\n{bank}:")
    print(bank_df['identified_themes'].value_counts().head(5))
```
